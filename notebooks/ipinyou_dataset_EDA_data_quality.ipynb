```python
compare_cols = [c for c in df_duplicated.columns if c != "bidid"]
results = []

# only process duplicated bidid groups 
```

```python
dup_groups = df_duplicated.groupby("bidid").filter(lambda g: len(g) > 1).groupby("bidid")
```

```python
for bidid, group in dup_groups:
    # Columns where values differ within this bidid
    diff_cols = [col for col in compare_cols if group[col].nunique(dropna=False) > 1]

    if diff_cols:
        # Store differing columns and their values
        diff_values = {
            col: group[col].tolist()
            for col in diff_cols
        }

        results.append({
            "bidid": bidid,
            "different_features": diff_cols,
            "values": diff_values
        })

# Convert to dataframe for easy viewing
diff_df = pd.DataFrame(results)
```

```python
#print(diff_df)

dict = {}

total_diff_in_timestamps = 0
num_diff_timestamps = 0

from datetime import datetime
# Parse: first 14 digits are yyyyMMddHHmmss, last 3 are milliseconds
def parse_ts(ts: str) -> datetime:
    return datetime.strptime(ts, '%Y%m%d%H%M%S%f')


# Optional: pretty print each bidid
for _, row in diff_df.iterrows():
    print(f"\nBidID: {row['bidid']}")
    print("Different features:", row["different_features"])
    for feature in row["different_features"]:
        if feature in dict:
            dict[feature] += 1
        else:
            dict[feature] = 1
    for feature, values in row["values"].items():
        print(f"  {feature}: {values}")
        if feature == "timestamp":
            #print(int(values[1])-int(values[0]))

            dt1 = parse_ts(values[0])
            dt2 = parse_ts(values[1]) 
            if len(values) > 2:
                print("number of timestamp values more than two")
                break
            delta_s = (dt2 - dt1).total_seconds()

            total_diff_in_timestamps += delta_s
            num_diff_timestamps += 1
```

Where the only difference between rows is the timestamp and ipinyouid, retain the row with the later timestamp

```python
import pandas as pd

# Make a copy
df_clean = df_duplicated.copy()

# Convert timestamp to datetime
df_clean["timestamp_dt"] = pd.to_datetime(
    df_clean["timestamp"].astype(str),
    format="%Y%m%d%H%M%S%f"
)

# Columns that must be identical for two rows to be considered duplicates
compare_cols = [
    col for col in df_clean.columns
    if col not in ["timestamp", "ipinyouid", "timestamp_dt"]
]

# Sort by timestamp so the newest row comes first
df_clean = df_clean.sort_values("timestamp_dt", ascending=False)

# Remove duplicates where every column except timestamp and iPinYouid
# is identical. The newest timestamp is retained.
df_clean = df_clean.drop_duplicates(
    subset=compare_cols,
    keep="first"
)

# Remove helper column
df_clean = df_clean.drop(columns="timestamp_dt")

# Reset index
df_clean = df_clean.reset_index(drop=True)
```

```python
print(f"The number of rows removed where the only difference was the timestamp \nand ipinyouid: {df_duplicated.shape[0]-df_clean.shape[0]}. The older timestamp record is retained.")
```

```python
import numpy as np

# Randomly select one row for each bidid
random_rows = df_duplicated.groupby("bidid").sample(n=1, random_state=42)

# Combine all usertags for each bidid
combined_tags = (
    df_duplicated.groupby("bidid")["usertag"]
      .apply(lambda x: ",".join(sorted(set(
          tag
          for value in x.dropna()
          for tag in value.split(",")
      ))))
)

# Replace the usertag in the randomly selected rows
random_rows["usertag"] = random_rows["bidid"].map(combined_tags)

# Result
df_combined = random_rows.reset_index(drop=True)
```

## Combine the cleaned duplicates with the original data

```python
# remove all original rows whose bidid has been tidied 
df_clean = df[~df['bidid'].isin(df_combined['bidid'])].copy()

# append the tidied duplicates back in 
df_clean = pd.concat(
    [df_clean, df_combined],
    ignore_index=True
)
```
